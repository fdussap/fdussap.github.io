### Instruction conditionnelle

Les instructions conditionnelles sont des instructions exécutées si une certaine condition (un booléen) est vrai. La syntaxe est la suivante :
```python
if condition:
    # code si la condition est vraie
    # ...
else:
    # code si la condition est fausse
    # ...
```
Python dispose également du mot-clé `elif` (contraction de *else if*), qui permet de tester une autre condition, si la condition précédente était fausse :
```python
if condition1:
    # code si condition1 est vraie
    # ...
elif condition2:
    # code si condition1 est fausse et condition2 est vraie
    # ...
else:
    # code si condition1 et condition2 sont fausses
    # ...
```

Le plus souvent, une condition sera une comparaison entre deux nombres :
- `a < b` : a est strictement inférieur à b
- `a <= b` : a est inférieur ou égal à b
- `a == b` : a est égal à b (ne pas confondre l'affectation `=` et le test d'égalité `==`)
- `a != b` : a est différent de b
  
On peut créer des conditions plus complexes avec les opérateurs logique `and` (et), `or` (ou) et `not` (négation).

### Boucles

La 1re boucle à connaitre est la boucle `for` (pour). Cette boucle répète ses instructions, pour une variable prenant les valeurs successives contenues dans une liste (plus exactement, dans un object itérable).
Syntaxe de la boucle `for` :
```python
for element in objet_iterable:
    # Intructions exécutées pour les valeurs successives contenues 
    # dans objet_iterable.
    # la variable element change de valeur à chaque tour de boucle.
```
Cette boucle est souvent utilisée avec la fonction `range`, qui crée un objet itérable contenant des entiers régulièrement espacés (consulter la documentation de cette fonction).

````python
# Executer ce code, puis changer range(10) en range(1,10), puis range(1,10,2)
for i in range(10):
    print(i)
````

```python
while condition:
    # Instructions exécutées en boucle.
    # La boucle ne s'arrête que si condition prend la valeur false.
```

### Tuples et listes

Les tuples et les listes sont des objets qui contiennent des éléments (qui peuvent être de types différents). La différence est que les tuples sont immutables (non modifiables) alors que les listes sont mutables (modifiables). On crée un tuple ou une liste comme ceci :

````python
a = (3.14, False, "bonjour")  # un tuple
b = [3.14, False, "bonjour"]  # une liste
````

Il est possible de créer une liste vide `[]`. Pour accéder à la valeur de l'élément en position $k$ d'une liste ou d'un tuple `l`, on écrit `l[k]`. En Python, les indices partent de 0 : le 1er élément est `l[0]`, le 2e élément est `l[1]`, etc. C'est également comme ça qu'on modifie la valeur d'un élément de la liste : `l[2] = 2026` attribue la valeur 2026 au 3e élément de la liste.

````python
a[2] = 2026 # produit une erreur car a est un tuple (supprimer la ligne ensuite)
b[2] = 2026
````

Les indices négatifs permettent d'accéder aux éléments en partant de la fin : `l[-1]` est le dernier élément de la liste, `l[-2]`est l'avant-dernier, etc.

#### Slicing

Le slicing permet d'extraire dans une **nouvelle** liste une partie d'une liste déjà existante. La syntaxe reprend celle de la fonction `range`.
```python
l[start:stop:step] # contient les éléments de l aux indices allant de start (inclus) jusqu à stop (exclu) avec un pas step
```
Si step est omis, il vaut 1 par défaut (comme pour range). Quelques cas particuliers :
- `l[i:]` : sélectionne les élément de $i$ jusqu'à la fin de la liste.
- `l[:i]` : sélectionne les éléments de 0 jusqu'à $i-1$.
- `l[::2]`: sélectionne un élément sur 2.
- `l[::-1]`: sélectionne tous les éléments, mais en parcourant la liste à l'envers.
- `l[-i:]`: sélectionne les $i$ derniers éléments.
- `l[:-i]`: sélectionne tous les éléments sauf les $i$ derniers.
- `l[:]` : sélectionne tout.

#### Fonctions sur les listes

Quelques fonctions à connaitre : `sum` (somme), `max`, `min`, `len` (*length*, la longueur) et `sorted` (crée une nouvelle liste triée dans l'ordre croissant, voir doc).
Pour une liste `l` de booléens :
- `any(l)` renvoie `True` ssi **au moins** l'un des éléments de la liste est vrai. 
- `all(l)` renvoie `True` ssi **tous** les éléments de la liste sont vrais.
- `sum(l)` compte le nombre d'éléments vrais.

Enfin, `+` permet de concaténer des listes. Attention, ce n'est pas l'addition terme à terme.

````python
[1, 2, 3] + ["bonjour", "bonsoir"]
````

Une liste est aussi un objet qui contient des **méthodes**, c'est-à-dire des fonctions attachées à la liste et qui agissent sur elle. Une méthode s'utilise avec la syntaxe :
```python 
nom_de_la_liste.nom_de_la_methode(...)
``` 
Quelques méthodes à connaitre pour les listes :
- `.append(x)` : ajoute `x` à la fin de la liste.
- `.sort()` : trie **en place** la liste dans l'ordre croissant (et d'autres options si vous lisez la doc).

Si vous écrivez `L2 = L1`, la liste `L2` n'est pas une copie indépendante de `L1`, mais est en fait une référence à la même liste : si vous modifiez `L1`, alors `L2` est modifiée aussi (car c'est la même liste).

#### Liste définie par compréhension

Une syntaxe utile pour créer des listes sans utiliser de boucle est la suivante.
```python
[expression for element in objet_iterable]
```
On crée une liste qui contient toutes les valeurs de `expression` lorsque `element` parcourt `objet_iterable`.
On notera le parallèle avec l'écriture ensembliste $\{ f(x) : x\in E \}$.

````python
# Exemple
[1/x for x in [1,2,3,4]]
````

On peut également ajouter une condition.
```python
[expression for element in objet_iterable if condition]
```

````python
# Exemple : anticiper le résulat
[n for n in [1,2,3,4] if n % 2 == 0]
````

Les fonctions `sum`, `max`, `min`, `all` et `any` fonctionnent aussi en compréhension. Par exemple, le code suivant calcule la somme des entiers pairs compris entre 1 et 100. 

````python
sum(n for n in range(1, 101) if n % 2 == 0)
````

### Indexation, slicing, masques

L'indexation des tableaux fonctionne exactement comme pour les listes, mais avec plusieurs indices si le tableau a plusieurs dimensions. Si `M` est une matrice :
- `M[i,j]` permet d'accéder au coefficient d'indices $i$, $j$ (attention, les indices partent toujours de 0)
- `M[i]` ou `M[i,:]` permet d'accéder à la $i$-ième ligne.
- `M[:,j]` permet d'accéder à la $j$-ième colonne.

Les tableaux NumPy sont mutables, comme pour les listes. Si vous voulez une copie indépendante d'un tableau, utiliser la fonction `np.copy()` ou la méthode `.copy()`.

On peut sélectionner plusieurs lignes/colonnes avec du slicing. 
**Attention :** contrairement aux listes, le slicing des tableaux NumPy ne renvoie pas une copie des valeurs, mais une vue sur le tableau de départ.

````python
# Exemple : modifier A modifie aussi M
M = np.arange(1, 13).reshape(3, 4)
A = M[:2, :3]  # les 2 premières lignes et les 3 premières colonnes de M
A[0,0] = 2026
print(A)
print(M)
````

Une autre façon de sélectionner des éléments d'un tableau est d'utiliser un **masque**. Un masque est un tableau de booléen `cond` de même taille que le tableau `M`.
- L'expression `M[cond]` sélectionne tous les coefficient à la même position qu'un `True` dans `cond`.
- L'instruction `M[cond] = a` assigne la valeur `a` aux positions qui correspondent à `True` dans `cond`.

````python
# Exemple
M = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

cond = np.array([
    [True, False, False],
    [False, True, True]
])

print(M[cond])
M[cond] = 0
print(M)
````

Le plus souvent, le masque est obtenu par compaison des valeurs de M avec une valeur. Par exemple `M > 0` est un tableau booléen de même taille que M, dont les coefficients valent `True` si le coefficient correspondant dans M est strictement positif. L'expression `M[M>0]` sélectionne alors les valeurs strictement positives de M.

Pour combiner des masque booléens avec des opérateurs logiques, on utilise `&` (et), `|` (ou) et `~` (négation). Par exemple, `M[(M>0) & (M<10)]` extrait les valeus de M strictement comprises entre 0 et 10.

### Échelles logarithmiques

Lors des prochains TP, on sera amenés à déterminer la vitesse de convergence vers 0 de l'erreur $e_n$ d'une méthode numérique. Les 2 cas de figures qu'on rencontre le plus :
- $e_n \sim \dfrac{C}{n^\alpha}$ avec $\alpha>0$ (vitesse algébrique);
- $e_n \sim C q^n$ avec $0<q<1$ (vitesse exponentielle ou géométrique).

La constante $C$ n'est pas très importante, la vitesse est avant tout dictée par l'exposant $\alpha$ pour les vitesses algébriques, et la raison $q$ pour la vitesse géométrique.

Pour mieux voir la vitesse de convergence d'une suite, on la représente dans un repère logarithmique ou semi-logarithmique. Les fonctions `plt.xscale` et `plt.yscale` permettent de changer l'échelle des axes.
